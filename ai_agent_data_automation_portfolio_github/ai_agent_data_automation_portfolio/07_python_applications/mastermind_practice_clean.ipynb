```python
import random

class Mastermind:

    def __init__(self):

        # ---------------------------------
        # 실제 정답 생성
        # 중복 허용
        # ---------------------------------
        self.__answer = []

        while len(self.__answer) < 4:
            n = random.randint(0, 9)
            self.__answer.append(n)


        # ---------------------------------
        # 가능한 정답 10000개 생성
        # ---------------------------------
        self.__possible_answers = []

        for a in range(10):
            for b in range(10):
                for c in range(10):
                    for d in range(10):
                        self.__possible_answers.append([a, b, c, d])


        # 정상적인 추측 횟수
        self.__count = 0


    # =====================================
    # 두 숫자열을 비교하는 내부 메서드
    # =====================================
    def __judge(self, guess, answer):

        exact = 0
        misplaced = 0

        answer_left = []
        guess_left = []


        # 1단계 : 숫자와 자리 모두 일치
        for i in range(4):

            if guess[i] == answer[i]:
                exact += 1

            else:
                answer_left.append(answer[i])
                guess_left.append(guess[i])


        # 2단계 : 숫자는 있지만 위치가 다름
        for n in guess_left:

            if n in answer_left:
                misplaced += 1
                answer_left.remove(n)


        # 나머지는 불일치
        absent = 4 - exact - misplaced

        return exact, misplaced, absent


    # =====================================
    # 사용자의 숫자를 추측
    # =====================================
    def guessNum(self, guess):

        # ---------------------------------
        # 입력 검사
        # ---------------------------------
        if len(guess) != 4:
            return '"네 자리" 숫자를 입력해야 합니다.'


        for n in guess:

            if type(n) != int:
                return '"숫자"를 입력해야 합니다.'

            if n < 0 or n > 9:
                return '0부터 9까지의 숫자를 입력해야 합니다.'


        # 정상적인 입력만 시행 횟수 증가
        self.__count += 1


        # ---------------------------------
        # 실제 정답과 비교
        # ---------------------------------
        exact, misplaced, absent = self.__judge(
            guess,
            self.__answer
        )


        # ---------------------------------
        # 현재 결과와 모순되는 후보 제거
        # ---------------------------------
        new_possible = []

        for candidate in self.__possible_answers:

            c_exact, c_misplaced, c_absent = self.__judge(
                guess,
                candidate
            )

            if (c_exact == exact and
                c_misplaced == misplaced and
                c_absent == absent):

                new_possible.append(candidate)


        self.__possible_answers = new_possible


        # ---------------------------------
        # 기본 출력
        # ---------------------------------
        result = (
            f"완전 일치 : {exact}\n"
            f"위치 불일치 : {misplaced}\n"
            f"불일치 : {absent}"
        )


        # ---------------------------------
        # 정답이면 종료 메시지
        # ---------------------------------
        if exact == 4:

            result += (
                "\n정답입니다."
                f"\n총 시행 횟수 : {self.__count}회"
            )

            return result


        # =================================
        # 중복 가능성 힌트
        #
        # 2회차 이후,
        # 남은 후보의 80% 이상이
        # 중복 숫자를 포함할 때만 표시
        # =================================
        duplicate_count = 0

        for candidate in self.__possible_answers:

            if len(set(candidate)) < 4:
                duplicate_count += 1


        if len(self.__possible_answers) > 0:

            duplicate_ratio = (
                duplicate_count /
                len(self.__possible_answers)
            )

            if (self.__count >= 2 and
                duplicate_ratio >= 0.8):

                result += "\n※ 중복 가능성 있음"


        # =================================
        # 추론 진행 보조장치
        # =================================

        remain = len(self.__possible_answers)


        # 정답 후보가 이미 하나뿐
        if remain == 1:

            result += (
                "\n※ 현재까지의 정보만으로 "
                "정답을 확정할 수 있습니다."
            )


        # 후보가 아주 조금만 남음
        elif remain <= 5:

            result += (
                "\n※ 새로운 숫자를 찾기보다 "
                "숫자와 자리 관계를 다시 살펴보세요."
            )


        # 후보가 꽤 좁혀짐
        elif remain <= 10:

            result += (
                "\n※ 지금까지의 결과를 "
                "다시 조합해 볼 만합니다."
            )


        return result

game1 = Mastermind()

while True:
    try:
        num = input("네 자리 숫자를 입력하세요 : ")

        guess = []

        for n in num:
            guess.append(int(n))

        result = game1.guessNum(guess)
        print(result)

        if "정답입니다." in result:
            break

    except ValueError:
        print('"숫자"를 입력해야 합니다.')
```

### 상세 설명: 단일 거대 메서드 방식 (Monolithic Class)
이 방식은 하나의 `guessNum` 메서드 안에 게임의 핵심 로직을 모두 집어넣은 구조입니다.

*   **`__init__`**: 게임 시작 시 정답(`__answer`)을 생성하고, 가능한 모든 후보 10,000개(`__possible_answers`)를 리스트로 만듭니다.
*   **`__judge`**: 두 숫자 리스트를 비교하여 스트라이크(완전 일치)와 볼(숫자만 일치)의 개수를 계산하는 내부 도구입니다.
*   **`guessNum`**: 사용자의 입력을 받아 다음 과정을 한 번에 수행합니다.
    1. **입력 검증**: 4자리 숫자인지, 정수인지 확인합니다.
    2. **판정**: 실제 정답과 비교해 결과를 냅니다.
    3. **필터링**: 10,000개의 후보 중 현재 결과와 모순되는 숫자를 모두 제거합니다.
    4. **힌트**: 남은 후보들의 상태를 분석해 중복 가능성이나 정답 확정 여부를 알려줍니다.

```python
import random


# =====================================
# 1. 정답 생성
# =====================================
def make_answer():
    answer = []

    while len(answer) < 4:
        n = random.randint(0, 9)
        answer.append(n)

    return answer


# =====================================
# 2. 가능한 정답 10000개 생성
# =====================================
def make_possible_answers():
    possible_answers = []

    for a in range(10):
        for b in range(10):
            for c in range(10):
                for d in range(10):
                    possible_answers.append([a, b, c, d])

    return possible_answers


# =====================================
# 3. 입력값을 리스트로 변환
# =====================================
def convert_guess(num):
    guess = []

    for n in num:
        guess.append(int(n))

    return guess


# =====================================
# 4. 입력 검사
# =====================================
def check_guess(guess):

    if len(guess) != 4:
        return False

    for n in guess:
        if n < 0 or n > 9:
            return False

    return True


# =====================================
# 5. 두 숫자열 판정
# =====================================
def judge(guess, answer):

    exact = 0
    misplaced = 0

    answer_left = []
    guess_left = []

    # 완전 일치부터 검사
    for i in range(4):

        if guess[i] == answer[i]:
            exact += 1

        else:
            answer_left.append(answer[i])
            guess_left.append(guess[i])

    # 위치 불일치 검사
    for n in guess_left:

        if n in answer_left:
            misplaced += 1
            answer_left.remove(n)

    absent = 4 - exact - misplaced

    return exact, misplaced, absent


# =====================================
# 6. 가능한 정답 필터링
# =====================================
def filter_answers(possible_answers, guess,
                   exact, misplaced, absent):

    new_possible = []

    for candidate in possible_answers:

        c_exact, c_misplaced, c_absent = judge(
            guess,
            candidate
        )

        if (c_exact == exact and
            c_misplaced == misplaced and
            c_absent == absent):

            new_possible.append(candidate)

    return new_possible


# =====================================
# 7. 중복 가능성 힌트
# =====================================
def duplicate_hint(possible_answers, count):

    if count < 2:
        return ""

    duplicate_count = 0

    for candidate in possible_answers:

        if len(set(candidate)) < 4:
            duplicate_count += 1

    duplicate_ratio = (
        duplicate_count / len(possible_answers)
    )

    if duplicate_ratio >= 0.8:
        return "※ 중복 가능성 있음"

    return ""


# =====================================
# 8. 추론 진행 힌트
# =====================================
def reasoning_hint(possible_answers):

    remain = len(possible_answers)

    if remain == 1:
        return (
            "※ 현재까지의 정보만으로 "
            "정답을 확정할 수 있습니다."
        )

    elif remain <= 5:
        return (
            "※ 새로운 숫자를 찾기보다 "
            "숫자와 자리 관계를 다시 살펴보세요."
        )

    elif remain <= 10:
        return (
            "※ 지금까지의 결과를 "
            "다시 조합해 볼 만합니다."
        )

    return ""


# =====================================
# 9. 게임 실행
# =====================================

answer = make_answer()
possible_answers = make_possible_answers()

count = 0

while True:

    try:
        num = input("네 자리 숫자를 입력하세요 : ")

        if len(num) != 4:
            print('"네 자리" 숫자를 입력해야 합니다.')
            continue

        guess = convert_guess(num)

        if check_guess(guess) == False:
            print("0부터 9까지의 숫자를 입력해야 합니다.")
            continue

        count += 1

        # 판정
        exact, misplaced, absent = judge(
            guess,
            answer
        )

        # 후보 필터링
        possible_answers = filter_answers(
            possible_answers,
            guess,
            exact,
            misplaced,
            absent
        )

        # 판정 결과
        print("완전 일치 :", exact)
        print("위치 불일치 :", misplaced)
        print("불일치 :", absent)

        # 정답
        if exact == 4:
            print("정답입니다.")
            print("총 시행 횟수 :", count, "회")
            break

        # 중복 힌트
        hint1 = duplicate_hint(
            possible_answers,
            count
        )

        if hint1 != "":
            print(hint1)

        # 추론 진행 힌트
        hint2 = reasoning_hint(
            possible_answers
        )

        if hint2 != "":
            print(hint2)

    except ValueError:
        print('"숫자"를 입력해야 합니다.')
```

### 상세 설명: 절차 지향 방식 (Procedural Approach)
클래스를 사용하지 않고 독립적인 함수들을 나열하여 만든 방식입니다.

*   **독립적 함수들**: `make_answer`, `judge`, `filter_answers` 등 각 기능이 독립된 함수로 분리되어 있어 각 기능을 따로 테스트하기 좋습니다.
*   **메인 루프**: 코드 하단의 `while True` 문이 '컨트롤러' 역할을 하여 함수들을 순서대로 호출합니다.
*   **상태 관리**: `possible_answers`와 같은 변수들을 전역적으로 관리하며, 루프 안에서 직접 업데이트해야 합니다.

```python
import random


class Mastermind:

    def __init__(self):
        self.answer = self.make_answer()
        self.possible_answers = self.make_possible_answers()
        self.count = 0


    # 1. 실제 정답 생성
    def make_answer(self):
        answer = []

        while len(answer) < 4:
            n = random.randint(0, 9)
            answer.append(n)

        return answer


    # 2. 가능한 정답 10000개 생성
    def make_possible_answers(self):
        possible_answers = []

        for a in range(10):
            for b in range(10):
                for c in range(10):
                    for d in range(10):
                        possible_answers.append([a, b, c, d])

        return possible_answers


    # 3. 문자열 입력을 숫자 리스트로 변환
    def convert_guess(self, num):
        guess = []

        for n in num:
            guess.append(int(n))

        return guess


    # 4. 입력 검사
    def check_guess(self, guess):
        if len(guess) != 4:
            return '"네 자리" 숫자를 입력해야 합니다.'

        for n in guess:
            if n < 0 or n > 9:
                return '0부터 9까지의 숫자를 입력해야 합니다.'

        return 'OK'


    # 5. 두 숫자열 비교
    def judge(self, guess, answer):
        exact = 0
        misplaced = 0

        answer_left = []
        guess_left = []

        # 완전 일치 먼저 검사
        for i in range(4):
            if guess[i] == answer[i]:
                exact += 1
            else:
                answer_left.append(answer[i])
                guess_left.append(guess[i])

        # 위치 불일치 검사
        for n in guess_left:
            if n in answer_left:
                misplaced += 1
                answer_left.remove(n)

        absent = 4 - exact - misplaced

        return exact, misplaced, absent


    # 6. 후보 필터링
    def filter_answers(self, guess, exact, misplaced, absent):
        new_possible = []

        for candidate in self.possible_answers:
            c_exact, c_misplaced, c_absent = self.judge(guess, candidate)

            if (c_exact == exact and
                c_misplaced == misplaced and
                c_absent == absent):
                new_possible.append(candidate)

        self.possible_answers = new_possible


    # 7. 중복 가능성 힌트
    def duplicate_hint(self):
        if self.count < 2:
            return ""

        duplicate_count = 0

        for candidate in self.possible_answers:
            if len(set(candidate)) < 4:
                duplicate_count += 1

        if len(self.possible_answers) == 0:
            return ""

        duplicate_ratio = duplicate_count / len(self.possible_answers)

        if duplicate_ratio >= 0.8:
            return "※ 중복 가능성 있음"

        return ""


    # 8. 진행도 힌트
    def progress_hint(self):
        remain = len(self.possible_answers)

        if remain == 1:
            return "※ 현재까지의 정보만으로 정답을 확정할 수 있습니다."

        elif remain <= 5:
            return "※ 새로운 숫자를 찾기보다 숫자와 자리 관계를 다시 살펴보세요."

        elif remain <= 10:
            return "※ 지금까지의 결과를 다시 조합해 볼 만합니다."

        return ""


    # 9. 한 번의 추측을 처리하는 파이프라인
    def guessNum(self, num):

        # [입력] 문자열이면 변환
        if type(num) == str:
            try:
                guess = self.convert_guess(num)
            except ValueError:
                return '"숫자"를 입력해야 합니다.'
        else:
            guess = num


        # [검사]
        check_result = self.check_guess(guess)
        if check_result != 'OK':
            return check_result


        # [시행 횟수 증가]
        self.count += 1


        # [판정]
        exact, misplaced, absent = self.judge(guess, self.answer)


        # [후보 필터링]
        self.filter_answers(guess, exact, misplaced, absent)


        # [기본 결과]
        result = (
            f"완전 일치 : {exact}\n"
            f"위치 불일치 : {misplaced}\n"
            f"불일치 : {absent}"
        )


        # [정답 확인]
        if exact == 4:
            result += (
                "\n정답입니다."
                f"\n총 시행 횟수 : {self.count}회"
            )
            return result


        # [중복 힌트]
        dup_hint = self.duplicate_hint()
        if dup_hint != "":
            result += "\n" + dup_hint


        # [진행도 힌트]
        prog_hint = self.progress_hint()
        if prog_hint != "":
            result += "\n" + prog_hint


        return result

game1 = Mastermind()

while True:
    num = input("네 자리 숫자를 입력하세요 : ")
    print(game1.guessNum(num))
```

### 상세 설명: 리팩토링된 클래스 방식 (Refactored Class)
클래스를 사용하되, 로직을 아주 작은 단위의 메서드들로 쪼개어 가독성을 높인 방식입니다.

*   **메서드 세분화**: `guessNum` 하나에 다 넣지 않고 `convert_guess`, `check_guess`, `duplicate_hint` 등으로 기능을 쪼갰습니다.
*   **가독성**: `guessNum` 메서드 내부를 읽어보면 전체적인 흐름이 마치 이야기처럼 읽히도록 설계되었습니다.
*   **유지보수**: 특정 기능(예: 힌트 방식)만 수정하고 싶을 때 해당 메서드만 찾아 고치면 되므로 관리가 쉽습니다.

```python
import random


class Mastermind:

    def __init__(self):

        self.answer = []

        while len(self.answer) < 4:
            n = random.randint(0, 9)
            self.answer.append(n)


        self.possible_answers = []

        for a in range(10):
            for b in range(10):
                for c in range(10):
                    for d in range(10):
                        self.possible_answers.append([a, b, c, d])


        self.count = 0


    # 판정
    def judge(self, guess, answer):

        exact = 0
        misplaced = 0

        answer_left = []
        guess_left = []


        for i in range(4):

            if guess[i] == answer[i]:
                exact += 1

            else:
                answer_left.append(answer[i])
                guess_left.append(guess[i])


        for n in guess_left:

            if n in answer_left:
                misplaced += 1
                answer_left.remove(n)


        absent = 4 - exact - misplaced

        return exact, misplaced, absent


    # 후보 제거
    def filter_answers(self, guess, exact, misplaced, absent):

        new_possible = []

        for candidate in self.possible_answers:

            c_exact, c_misplaced, c_absent = self.judge(
                guess,
                candidate
            )

            if (c_exact == exact and
                c_misplaced == misplaced and
                c_absent == absent):

                new_possible.append(candidate)


        self.possible_answers = new_possible


    # 한 회차 실행
    def guessNum(self, guess):

        if len(guess) != 4:
            return '"네 자리" 숫자를 입력해야 합니다.'


        self.count += 1


        exact, misplaced, absent = self.judge(
            guess,
            self.answer
        )


        self.filter_answers(
            guess,
            exact,
            misplaced,
            absent
        )


        result = (
            f"완전 일치 : {exact}\n"
            f"위치 불일치 : {misplaced}\n"
            f"불일치 : {absent}"
        )


        if exact == 4:

            result += (
                "\n정답입니다."
                f"\n총 시행 횟수 : {self.count}회"
            )

            return result


        # 중복 힌트
        duplicate_count = 0

        for candidate in self.possible_answers:

            if len(set(candidate)) < 4:
                duplicate_count += 1


        if self.count >= 2:

            duplicate_ratio = (
                duplicate_count /
                len(self.possible_answers)
            )

            if duplicate_ratio >= 0.8:
                result += "\n※ 중복 가능성 있음"


        # 진행도 힌트
        remain = len(self.possible_answers)

        if remain == 1:

            result += (
                "\n※ 현재까지의 정보만으로 "
                "정답을 확정할 수 있습니다."
            )

        elif remain <= 5:

            result += (
                "\n※ 새로운 숫자를 찾기보다 "
                "숫자와 자리 관계를 다시 살펴보세요."
            )

        elif remain <= 10:

            result += (
                "\n※ 지금까지의 결과를 "
                "다시 조합해 볼 만합니다."
            )


        return result

game1 = Mastermind()

while True:

    try:
        num = input("네 자리 숫자를 입력하세요 : ")

        guess = []

        for n in num:
            guess.append(int(n))


        result = game1.guessNum(guess)

        print(result)


        if "정답입니다." in result:
            break


    except ValueError:
        print('"숫자"를 입력해야 합니다.')
```

### 상세 설명: 기본 클래스 방식 (Basic Class Approach)
가장 표준적인 형태의 클래스 구현 방식입니다.

*   **속성(Attributes)**: `self.answer`, `self.possible_answers`, `self.count`를 사용하여 게임의 상태를 객체 내부에 저장합니다.
*   **메서드(Methods)**: `judge`(판정), `filter_answers`(후보 제거), `guessNum`(추측 처리)으로 기능을 분리하여 클래스를 정의했습니다.
*   **특징**: `__init__`에서 모든 초기화를 직접 수행하며, 메서드들이 서로의 데이터를 공유하도록 설계되었습니다. 리팩토링 전의 표준적인 객체 지향 입문용 구조입니다.

```python
import random


class Mastermind:

    def __init__(self):

        self.__answer = self.__make_answer()
        self.__possible_answers = self.__make_possible_answers()
        self.__count = 0


    # 실제 정답 생성
    def __make_answer(self):

        answer = []

        while len(answer) < 4:
            n = random.randint(0, 9)
            answer.append(n)

        return answer


    # 가능한 정답 10000개 생성
    def __make_possible_answers(self):

        possible_answers = []

        for a in range(10):
            for b in range(10):
                for c in range(10):
                    for d in range(10):
                        possible_answers.append([a, b, c, d])

        return possible_answers


    # 문자열 → 숫자 리스트
    def __convert_guess(self, num):

        guess = []

        for n in num:
            guess.append(int(n))

        return guess


    # 입력 검사
    def __check_guess(self, guess):

        if len(guess) != 4:
            return '"네 자리" 숫자를 입력해야 합니다.'

        for n in guess:

            if type(n) != int:
                return '"숫자"를 입력해야 합니다.'

            if n < 0 or n > 9:
                return '0부터 9까지의 숫자를 입력해야 합니다.'

        return "OK"


    # 두 숫자열 판정
    def __judge(self, guess, answer):

        exact = 0
        misplaced = 0

        answer_left = []
        guess_left = []

        # 완전 일치 먼저 검사
        for i in range(4):

            if guess[i] == answer[i]:
                exact += 1

            else:
                answer_left.append(answer[i])
                guess_left.append(guess[i])


        # 위치 불일치 검사
        for n in guess_left:

            if n in answer_left:
                misplaced += 1
                answer_left.remove(n)


        absent = 4 - exact - misplaced

        return exact, misplaced, absent


    # 후보 필터링
    def __filter_answers(self, guess, exact, misplaced, absent):

        new_possible = []

        for candidate in self.__possible_answers:

            c_exact, c_misplaced, c_absent = self.__judge(
                guess,
                candidate
            )

            if (c_exact == exact and
                c_misplaced == misplaced and
                c_absent == absent):

                new_possible.append(candidate)


        self.__possible_answers = new_possible


    # 중복 힌트
    def __duplicate_hint(self):

        if self.__count < 2:
            return ""

        duplicate_count = 0

        for candidate in self.__possible_answers:

            if len(set(candidate)) < 4:
                duplicate_count += 1


        if len(self.__possible_answers) == 0:
            return ""


        duplicate_ratio = (
            duplicate_count /
            len(self.__possible_answers)
        )


        if duplicate_ratio >= 0.8:
            return "※ 중복 가능성 있음"

        return ""


    # 진행도 힌트
    def __progress_hint(self):

        remain = len(self.__possible_answers)

        if remain == 1:
            return (
                "※ 현재까지의 정보만으로 "
                "정답을 확정할 수 있습니다."
            )

        elif remain <= 5:
            return (
                "※ 새로운 숫자를 찾기보다 "
                "숫자와 자리 관계를 다시 살펴보세요."
            )

        elif remain <= 10:
            return (
                "※ 지금까지의 결과를 "
                "다시 조합해 볼 만합니다."
            )

        return ""


    # 사용자가 실제로 호출하는 메서드
    def guessNum(self, num):

        if type(num) == str:

            try:
                guess = self.__convert_guess(num)

            except ValueError:
                return '"숫자"를 입력해야 합니다.'

        else:
            guess = num


        check_result = self.__check_guess(guess)

        if check_result != "OK":
            return check_result


        self.__count += 1


        exact, misplaced, absent = self.__judge(
            guess,
            self.__answer
        )


        self.__filter_answers(
            guess,
            exact,
            misplaced,
            absent
        )


        result = (
            f"완전 일치 : {exact}\n"
            f"위치 불일치 : {misplaced}\n"
            f"불일치 : {absent}"
        )


        if exact == 4:

            result += (
                "\n정답입니다."
                f"\n총 시행 횟수 : {self.__count}회"
            )

            return result


        duplicate_hint = self.__duplicate_hint()

        if duplicate_hint != "":
            result += "\n" + duplicate_hint


        progress_hint = self.__progress_hint()

        if progress_hint != "":
            result += "\n" + progress_hint


        return result

game1 = Mastermind()

while True:

    num = input("네 자리 숫자를 입력하세요 : ")

    result = game1.guessNum(num)

    print(result)

    if "정답입니다." in result:
        break
```

### 상세 설명: 캡슐화 강조 방식 (Encapsulated Class)
정보 은닉과 입력의 유연성에 초점을 맞춘 방식입니다.

*   **캡슐화(`__`)**: 변수와 메서드 이름 앞에 언더바 두 개를 붙여 외부에서 정답이나 후보 리스트를 마음대로 조작할 수 없게 보호했습니다.
*   **입력 유연성**: 사용자가 문자열을 넣든 리스트를 넣든 `guessNum`이 스스로 판단하여 적절히 변환하여 처리합니다.

```python
import random


class Mastermind:

    def __init__(self):

        self.answer = []

        while len(self.answer) < 4:
            n = random.randint(0, 9)
            self.answer.append(n)


        self.possible_answers = []

        for a in range(10):
            for b in range(10):
                for c in range(10):
                    for d in range(10):
                        self.possible_answers.append([a, b, c, d])


        self.count = 0


    # ----------------------------------
    # 판정
    # ----------------------------------
    def judge(self, guess, answer):

        exact = 0
        misplaced = 0

        answer_left = []
        guess_left = []


        for i in range(4):

            if guess[i] == answer[i]:
                exact += 1

            else:
                answer_left.append(answer[i])
                guess_left.append(guess[i])


        for n in guess_left:

            if n in answer_left:
                misplaced += 1
                answer_left.remove(n)


        absent = 4 - exact - misplaced

        return exact, misplaced, absent


    # ----------------------------------
    # 후보 필터링
    # ----------------------------------
    def filter_answers(self, guess, exact, misplaced, absent):

        new_possible = []

        for candidate in self.possible_answers:

            c_exact, c_misplaced, c_absent = self.judge(
                guess,
                candidate
            )

            if (c_exact == exact and
                c_misplaced == misplaced and
                c_absent == absent):

                new_possible.append(candidate)


        self.possible_answers = new_possible


    # ----------------------------------
    # 중복 힌트
    # ----------------------------------
    def duplicate_hint(self):

        if self.count < 2:
            return ""


        duplicate_count = 0

        for candidate in self.possible_answers:

            if len(set(candidate)) < 4:
                duplicate_count += 1


        duplicate_ratio = (
            duplicate_count /
            len(self.possible_answers)
        )


        if duplicate_ratio >= 0.8:
            return "※ 중복 가능성 있음"

        return ""


    # ----------------------------------
    # 진행 힌트
    # ----------------------------------
    def progress_hint(self):

        remain = len(self.possible_answers)


        if remain == 1:

            return (
                "※ 현재까지의 정보만으로 "
                "정답을 확정할 수 있습니다."
            )


        elif remain <= 5:

            return (
                "※ 새로운 숫자를 찾기보다 "
                "숫자와 자리 관계를 다시 살펴보세요."
            )


        elif remain <= 10:

            return (
                "※ 지금까지의 결과를 "
                "다시 조합해 볼 만합니다."
            )


        return ""


    # ----------------------------------
    # 한 번의 추측
    # ----------------------------------
    def guessNum(self, guess):

        self.count += 1


        exact, misplaced, absent = self.judge(
            guess,
            self.answer
        )


        self.filter_answers(
            guess,
            exact,
            misplaced,
            absent
        )


        result = (
            f"완전 일치 : {exact}\n"
            f"위치 불일치 : {misplaced}\n"
            f"불일치 : {absent}"
        )


        if exact == 4:

            result += (
                "\n정답입니다."
                f"\n총 시행 횟수 : {self.count}회"
            )

            return result


        hint1 = self.duplicate_hint()

        if hint1 != "":
            result += "\n" + hint1


        hint2 = self.progress_hint()

        if hint2 != "":
            result += "\n" + hint2


        return result


    # ----------------------------------
    # 실제 게임 전체 실행
    # ----------------------------------
    def play(self):

        while True:

            try:

                num = input("네 자리 숫자를 입력하세요 : ")


                if len(num) != 4:
                    print('"네자리" 숫자를 입력해야 합니다.')
                    continue


                guess = []

                for n in num:
                    guess.append(int(n))


                result = self.guessNum(guess)

                print(result)


                if "정답입니다." in result:
                    break


            except ValueError:

                print('"숫자"를 입력해야 합니다.')

game1 = Mastermind()
game1.play()
```

### 상세 설명: 완전 자동화 클래스 방식 (Fully Automated Class)
게임 실행 로직까지 클래스 내부로 포함시킨 형태입니다.

*   **`play()` 메서드**: 게임을 돌리는 `while` 루프 자체가 클래스의 기능이 되었습니다. 사용자는 `game.play()` 한 줄만 실행하면 됩니다.
*   **코드 구획**: 주석(`----`)을 사용하여 판정, 필터링, 힌트, 실행 영역을 명확히 구분하여 구조적 안정감을 줍니다.
